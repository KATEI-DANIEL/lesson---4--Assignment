```python
# Python and Pandas Assessment Notebook

import pandas as pd
import numpy as np
from functools import reduce

# Task 1: Create a DataFrame using Pandas
# Create a DataFrame with the following data:
# Name: Alice, Bob, Charlie, David
# Age: 25, 30, 35, 28
# City: New York, San Francisco, Los Angeles, Chicago

# Create a dictionary containing the data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
}

# Create a DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
 # Task 2: Row and Column Manipulation
# Drop the 'City' column from the DataFrame created in Task 1

df = df.drop(columns=['City'])

# Display the DataFrame after dropping the 'City' column
print(df)
```

```python
# Task 3: Handling Null Values
# Create a new DataFrame with some null values and fill them
# Create a DataFrame with null values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, None, 35, 28],
    'Salary': [50000, 60000, None, 45000]
}

df = pd.DataFrame(data)
print(df)

# Fill null values with specified values
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

# Display the updated DataFrame
print("\nDataFrame after Filling Null Values:")
print(df)
```

```python
# Task 5: Concatenation and Merging
# Concatenate two DataFrames vertically and merge them horizontally

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Concatenate df1 and df2 vertically
df_concat = pd.concat([df1, df2], axis=0)

# Display the result

print('\nConcatenate df1 and df2 vertically')
print(df_concat)

# Merge df1 and df3 horizontally
df_merged = pd.concat([df1, df3], axis=1)

# Display the result
print('\nMerge df1 and df3 horizontally')
print(df_merged)
```

```python
# Task 8: Functions and Lambda
# Create a function to calculate the square of a number.
# Then, use a lambda function to do the same.

# Create a function to calculate the square of a number
def square(number):
    return number ** 2

# Call the function
result = square(5)

# Display the result
print("Square using a normal function:", result)
```

```python
# Create a list of the first 5 even numbers
even_numbers = [2, 4, 6, 8, 10]

# Create an iterator
even_iterator = iter(even_numbers)

# Display the numbers using next()
print("First 5 even numbers using an Iterator:")

print(next(even_iterator))
print(next(even_iterator))
print(next(even_iterator))
print(next(even_iterator))
print(next(even_iterator))
```

```python
# Create a generator function
def even_generator():
    for number in range(2, 11, 2):
        yield number

# Call the generator
even_numbers = even_generator()

# Display the numbers
print("First 5 even numbers using a Generator:")

for number in even_numbers:
    print(number)
```

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]
# Square all numbers using map()
squared_numbers = list(map(lambda x: x ** 2, numbers))

# Display the result
print("Squared numbers:", squared_numbers)

from functools import reduce

# Find the product using reduce()
product = reduce(lambda x, y: x * y, numbers)

# Display the result
print("Product of all numbers:", product)

#Filter even numbers
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

# Display the result
print("Even numbers:", even_numbers)
```

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Create a Rectangle class
class Rectangle:

    # Initialize the attributes
    def __init__(self, length, width):
        self.length = length
        self.width = width

    # Method to calculate area
    def calculate_area(self):
        return self.length * self.width

    # Method to calculate perimeter
    def calculate_perimeter(self):
        return 2 * (self.length + self.width)


# Create an object of the Rectangle class
rectangle1 = Rectangle(10, 5)

# Calculate and display area and perimeter
print("Length:", rectangle1.length)
print("Width:", rectangle1.width)
print("Area:", rectangle1.calculate_area())
print("Perimeter:", rectangle1.calculate_perimeter())
```

```python
# Task 12: Pandas Data Analysis
# Using the following DataFrame, perform these tasks:
# 1. Find the average salary by department
# 2. Get the names of employees with salary > 60000
# 3. Add a new column 'Bonus' which is 10% of salary

df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# Calculate average salary by department
average_salary = df_employees.groupby('Department')['Salary'].mean()

print("Average Salary by Department:")
print(average_salary)

# Filter employees earning more than 60000
high_salary = df_employees[df_employees['Salary'] > 60000]['Name']

print("Employees earning more than 60000:")
print(high_salary)

# Calculate bonus (10% of salary)
df_employees['Bonus'] = df_employees['Salary'] * 0.10

# Display updated DataFrame
print("Updated Employee DataFrame:")
print(df_employees)

```
